## One demo of each tool

`toolkit_demo.py`

```python
"""Lesson 1.4 - one short demo of each tool in the Python toolkit."""

import heapq
from bisect import bisect_left
from collections import Counter, defaultdict, deque

# 1. Counter: how many times does each item appear?
count = Counter("banana")
print("Counter          :", dict(count))              # {'b': 1, 'a': 3, 'n': 2}
print("most_common(1)   :", count.most_common(1))    # [('a', 3)]

# 2. defaultdict: grouping without checking if the key exists
groups = defaultdict(list)
for word in ["cat", "dog", "bird", "fish", "ox"]:
    groups[len(word)].append(word)
print("defaultdict      :", dict(groups))

# 3. deque: O(1) at both ends
q = deque([1, 2, 3])
q.append(4)
first = q.popleft()
print("deque            :", first, list(q))

# 4. heapq: a min heap, push and pop are O(log n)
h = []
for x in [5, 1, 8, 3]:
    heapq.heappush(h, x)
print("heap pops        :", [heapq.heappop(h) for _ in range(len(h))])

# 5. bisect: binary search on a sorted list, O(log n)
print("bisect_left      :", bisect_left([1, 3, 5, 7], 5))      # 2

# 6. sorted with key: sort by any rule you define
students = [("Asha", 82), ("Ravi", 90), ("Meena", 75)]
print("sorted by marks  :", sorted(students, key=lambda s: s[1], reverse=True))

# Small features that save time
for index, value in enumerate(["a", "b", "c"]):
    print("enumerate        :", index, value)
for x, y in zip([1, 2, 3], ["one", "two", "three"]):
    print("zip              :", x, y)
a, b = 1, 2
a, b = b, a
print("swap             :", a, b)
smallest = float("inf")
for x in [7, 3, 9]:
    smallest = min(smallest, x)
print("float('inf') min :", smallest)
point = (4, 5)
px, py = point
print("tuple unpacking  :", px, py)
```

## Warm-up 1: Counter

`count_vowels_counter.py`

```python
"""Lesson 1.4 - Warm-up 1: count the vowels in a string using Counter."""

from collections import Counter

VOWELS = "aeiou"


def count_vowels(text):
    """Time O(n), space O(1) (at most 26 distinct letters)."""
    counts = Counter(text.lower())
    return sum(counts[v] for v in VOWELS)


def vowel_breakdown(text):
    """Return how many times each vowel appears."""
    counts = Counter(text.lower())
    return {v: counts[v] for v in VOWELS if counts[v] > 0}


if __name__ == "__main__":
    for text in ["banana", "Hello World", "rhythm", "", "AEIOU aeiou"]:
        print(repr(text), "->", count_vowels(text), vowel_breakdown(text))
```

## Warm-up 2: defaultdict

`group_words_by_length.py`

```python
"""Lesson 1.4 - Warm-up 2: group a list of words by length using defaultdict."""

from collections import defaultdict


def group_by_length(words):
    """Time O(n), space O(n). Returns {length: [words...]}."""
    groups = defaultdict(list)
    for word in words:
        groups[len(word)].append(word)
    return dict(groups)


if __name__ == "__main__":
    words = ["cat", "dog", "bird", "fish", "ox", "tiger", "bee"]
    print(group_by_length(words))
    # {3: ['cat', 'dog', 'bee'], 4: ['bird', 'fish'], 2: ['ox'], 5: ['tiger']}
    print(group_by_length([]))
```

## Warm-up 3: bisect_left

`bisect_insert_position.py`

```python
"""Lesson 1.4 - Warm-up 3: where should 40 be inserted in a sorted list?"""

from bisect import bisect_left


def insert_position(sorted_list, value):
    """Index where value belongs to keep the list sorted. Time O(log n)."""
    return bisect_left(sorted_list, value)


if __name__ == "__main__":
    data = [10, 20, 30, 50, 60]
    pos = insert_position(data, 40)
    print("40 goes at index", pos)          # 3

    data.insert(pos, 40)                    # insert itself is O(n), finding is O(log n)
    print("after insert:", data)            # [10, 20, 30, 40, 50, 60]

    # Edge cases
    print(insert_position([], 5))                 # 0
    print(insert_position([10, 20], 5))           # 0 (before everything)
    print(insert_position([10, 20], 99))          # 2 (after everything)
    print(insert_position([10, 20, 20, 30], 20))  # 1 (leftmost of equal values)
```

## Warm-up 4: sorted with key

`sort_by_marks_then_name.py`

```python
"""Lesson 1.4 - Warm-up 4: sort (name, marks) pairs by marks descending,
then by name ascending when the marks are equal."""


def sort_by_marks_then_name(students):
    """Time O(n log n). The key is a tuple: (-marks, name)."""
    return sorted(students, key=lambda s: (-s[1], s[0]))


if __name__ == "__main__":
    students = [
        ("Ravi", 90),
        ("Asha", 82),
        ("Meena", 90),
        ("Kiran", 82),
        ("Zoya", 75),
    ]
    for name, marks in sort_by_marks_then_name(students):
        print(f"{name:<8}{marks}")
    # Meena and Ravi tie on 90 -> Meena first (name ascending)
    # Asha and Kiran tie on 82 -> Asha first
```

## Warm-up 5: min and max in one pass

`min_max_single_pass.py`

```python
"""Lesson 1.4 - Warm-up 5: minimum and maximum in a single pass."""


def min_max(nums):
    """One loop, two variables. Time O(n), space O(1).

    Starts from float("inf") and float("-inf") so any real number beats them.
    Raises ValueError for an empty list.
    """
    if not nums:
        raise ValueError("list is empty")
    smallest = float("inf")
    biggest = float("-inf")
    for x in nums:
        if x < smallest:
            smallest = x
        if x > biggest:
            biggest = x
    return smallest, biggest


if __name__ == "__main__":
    tests = [[3, 9, 2], [-5, -1, -8], [7], [4, 4, 4], [1, 2, 3, 4, 5]]
    for nums in tests:
        assert min_max(nums) == (min(nums), max(nums))
        print(nums, "->", min_max(nums))
    try:
        min_max([])
    except ValueError as err:
        print("[] ->", err)
```
