## The 8 find-the-complexity snippets

`complexity_check_snippets.py`

```python
"""Lesson 1.2 - the 8 "find the complexity" snippets, runnable.

Each snippet returns the number of basic steps it performed, so you can
watch the growth yourself instead of only reading the answer.
"""


def snippet_1(n):                      # O(n)
    steps = 0
    for i in range(n):
        steps += 1
    return steps


def snippet_2(n):                      # O(n^2)
    steps = 0
    for i in range(n):
        for j in range(n):
            steps += 1
    return steps


def snippet_3(n):                      # O(n^2)  (0+1+...+(n-1) = n(n-1)/2)
    steps = 0
    for i in range(n):
        for j in range(i):
            steps += 1
    return steps


def snippet_4(n):                      # O(log n)  (i doubles each time)
    steps = 0
    i = 1
    while i < n:
        i *= 2
        steps += 1
    return steps


def snippet_5(n):                      # O(n)  (sequential loops add: n + n)
    steps = 0
    for i in range(n):
        steps += 1
    for j in range(n):
        steps += 1
    return steps


def snippet_6(n):                      # O(n)  (inner loop is fixed at 10)
    steps = 0
    for i in range(n):
        for j in range(10):
            steps += 1
    return steps


def snippet_7(n):                      # O(2^n)  (each call makes two calls)
    if n <= 1:
        return 1
    return snippet_7(n - 1) + snippet_7(n - 1)


def snippet_8(n):                      # O(n log n)  (outer n, inner log n)
    steps = 0
    for i in range(n):
        j = 1
        while j < n:
            j *= 2
            steps += 1
    return steps


ANSWERS = {
    1: "O(n)",
    2: "O(n^2)",
    3: "O(n^2)",
    4: "O(log n)",
    5: "O(n)",
    6: "O(n)",
    7: "O(2^n)",
    8: "O(n log n)",
}

if __name__ == "__main__":
    sizes = [10, 100, 1000]
    print(f"{'snippet':<10}{'answer':<12}" + "".join(f"n={n:<10}" for n in sizes))
    for k, func in enumerate(
        [snippet_1, snippet_2, snippet_3, snippet_4,
         snippet_5, snippet_6, snippet_8], start=1
    ):
        number = k if k < 7 else 8
        row = "".join(f"{func(n):<12}" for n in sizes)
        print(f"{number:<10}{ANSWERS[number]:<12}{row}")

    print()
    print("Snippet 7 is exponential, so use small n only:")
    for n in (5, 10, 15, 20):
        print(f"  n={n:<3} -> {snippet_7(n)} leaf calls")
```

## Practice sheet: 8 more snippets

`complexity_practice_snippets.py`

```python
"""Lesson 1.2 - Practice sheet: find the complexity of 8 more snippets.

Read each function, write your answer on paper, THEN scroll to ANSWERS.
a and b are the sizes of two different inputs; n is the size of one input.
"""


def practice_1(n):
    count = 0
    for i in range(n):
        for j in range(n):
            for k in range(n):
                count += 1
    return count


def practice_2(n):
    count = 0
    for i in range(0, n, 2):
        count += 1
    return count


def practice_3(n):
    count = 0
    while n > 1:
        n //= 2
        count += 1
    return count


def practice_4(list_a, list_b):
    count = 0
    for x in list_a:
        count += 1
    for y in list_b:
        count += 1
    return count


def practice_5(list_a, list_b):
    count = 0
    for x in list_a:
        for y in list_b:
            count += 1
    return count


def practice_6(nums):
    seen = []
    for x in nums:
        if x in seen:          # hidden loop: "in" on a list
            return True
        seen.append(x)
    return False


def practice_7(nums):
    nums = sorted(nums)        # hidden cost of sorted()
    total = 0
    for x in nums:
        total += x
    return total


def practice_8(nums):
    return nums[0] + nums[-1]


ANSWERS = {
    1: ("O(n^3)", "Three nested loops, each n."),
    2: ("O(n)", "Step of 2 is still a constant factor: n/2 -> n."),
    3: ("O(log n)", "n is halved each time."),
    4: ("O(a + b)", "Sequential loops add; different inputs, different letters."),
    5: ("O(a * b)", "Nested loops multiply."),
    6: ("O(n^2)", "Loop of n, and 'x in seen' scans up to n items each time."),
    7: ("O(n log n)", "sorted() is n log n; the loop is only n, so sorting wins."),
    8: ("O(1)", "Two index lookups, no loop."),
}

if __name__ == "__main__":
    print("Sanity run (the functions do work):")
    print("practice_1(10)      =", practice_1(10))
    print("practice_2(10)      =", practice_2(10))
    print("practice_3(1000)    =", practice_3(1000))
    print("practice_4(3, 4 items) =", practice_4([1, 2, 3], [1, 2, 3, 4]))
    print("practice_5(3, 4 items) =", practice_5([1, 2, 3], [1, 2, 3, 4]))
    print("practice_6([1,2,3,1]) =", practice_6([1, 2, 3, 1]))
    print("practice_7([3,1,2]) =", practice_7([3, 1, 2]))
    print("practice_8([5,6,7]) =", practice_8([5, 6, 7]))
    print()
    print("Answers:")
    for k, (big_o, why) in ANSWERS.items():
        print(f"  {k}. {big_o:<11} {why}")
```

## Will it pass or time out?

`does_it_pass_estimator.py`

```python
"""Lesson 1.2 - Practice: will my solution pass or time out?

Rule of thumb from the lesson: Python does about 10^7 simple operations per
second. Change TIME_LIMIT_SECONDS to match the platform you practise on.
"""

import math

OPS_PER_SECOND = 10**7
TIME_LIMIT_SECONDS = 3

COMPLEXITIES = {
    "O(log n)": lambda n: math.log2(n),
    "O(n)": lambda n: n,
    "O(n log n)": lambda n: n * math.log2(n),
    "O(n^2)": lambda n: n**2,
    "O(n^3)": lambda n: n**3,
}


def estimate(complexity, n):
    """Return (steps, seconds, passes) for a complexity at input size n."""
    steps = COMPLEXITIES[complexity](n)
    seconds = steps / OPS_PER_SECOND
    return steps, seconds, seconds <= TIME_LIMIT_SECONDS


def show(n_values):
    print(f"Limit: {TIME_LIMIT_SECONDS}s at {OPS_PER_SECOND:.0e} ops/second\n")
    for n in n_values:
        print(f"n = {n:,}")
        for name in COMPLEXITIES:
            steps, seconds, ok = estimate(name, n)
            verdict = "PASS" if ok else "TIMEOUT"
            print(f"  {name:<11} ~{steps:>10.2e} steps  ~{seconds:>12.4f} s  {verdict}")
        print()


if __name__ == "__main__":
    # The three sizes from the lesson practice: 10^4, 10^5 and 10^6
    show([10**4, 10**5, 10**6])
```
