## Example A: find the maximum in a list

`find_max.py`

```python
"""Lesson 1.1 - Example A: find the maximum in a list."""


def find_max_brute_force(nums):
    """Compare every element with every other element.

    Time O(n^2), space O(1).
    """
    if not nums:
        raise ValueError("list is empty")
    for i in range(len(nums)):
        is_biggest = True
        for j in range(len(nums)):
            if nums[j] > nums[i]:
                is_biggest = False
                break
        if is_biggest:
            return nums[i]


def find_max(nums):
    """Keep a running maximum and scan once.

    Time O(n), space O(1).
    """
    if not nums:
        raise ValueError("list is empty")
    best = nums[0]
    for x in nums:
        if x > best:
            best = x
    return best


if __name__ == "__main__":
    tests = [
        [3, 9, 2],          # normal
        [-5, -1, -8],       # all negative
        [7],                # single element
        [4, 4, 4],          # all equal
        [1, 2, 3, 4, 5],    # already sorted
        [5, 4, 3, 2, 1],    # reverse sorted
    ]
    for nums in tests:
        assert find_max(nums) == find_max_brute_force(nums) == max(nums)
        print(nums, "->", find_max(nums))

    try:
        find_max([])
    except ValueError as err:
        print("[] ->", err)
```

## Example B: check whether a list is sorted

`is_sorted.py`

```python
"""Lesson 1.1 - Example B: check whether a list is sorted (ascending)."""


def is_sorted_brute_force(nums):
    """Sort a copy and compare.

    Time O(n log n), extra space O(n).
    """
    return nums == sorted(nums)


def is_sorted(nums):
    """Scan once and stop at the first violation.

    Time O(n), space O(1).
    """
    for i in range(1, len(nums)):
        if nums[i] < nums[i - 1]:
            return False
    return True


if __name__ == "__main__":
    tests = [
        ([1, 2, 3, 4], True),
        ([1, 3, 2, 4], False),
        ([], True),             # empty
        ([5], True),            # single element
        ([2, 2, 2], True),      # all equal
        ([5, 4, 3], False),     # reverse sorted
        ([-3, -1, 0, 2], True), # negatives and zero
    ]
    for nums, expected in tests:
        assert is_sorted(nums) == is_sorted_brute_force(nums) == expected
        print(nums, "->", is_sorted(nums))
```

## Practice: largest and second largest

`largest_and_second_largest.py`

```python
"""Lesson 1.1 - Practice: find the largest and second largest in a list.

"Second largest" here means the next smaller DISTINCT value.
If there is no such value (single element, all equal), it is None.
"""


def largest_and_second_largest_brute_force(nums):
    """Sort the distinct values. Time O(n log n), space O(n)."""
    if not nums:
        raise ValueError("list is empty")
    unique = sorted(set(nums), reverse=True)
    second = unique[1] if len(unique) > 1 else None
    return unique[0], second


def largest_and_second_largest(nums):
    """One pass, two variables. Time O(n), space O(1)."""
    if not nums:
        raise ValueError("list is empty")
    largest = float("-inf")
    second = float("-inf")
    for x in nums:
        if x > largest:
            second = largest
            largest = x
        elif largest > x > second:
            second = x
    if second == float("-inf"):
        second = None
    return largest, second


if __name__ == "__main__":
    tests = [
        [3, 9, 2, 7],      # normal
        [-5, -1, -8],      # negatives
        [4, 4, 4],         # all equal -> no second largest
        [9],               # single element
        [1, 2, 3, 4, 5],   # sorted
        [5, 4, 3, 2, 1],   # reverse sorted
        [8, 8, 5, 5],      # duplicates
    ]
    for nums in tests:
        fast = largest_and_second_largest(nums)
        slow = largest_and_second_largest_brute_force(nums)
        assert fast == slow, (nums, fast, slow)
        print(nums, "->", fast)
```

## Practice: the 7 steps for reversing a list

`reverse_list_seven_steps.py`

```python
"""Lesson 1.1 - Practice: the 7 steps applied to "Reverse a List"."""

# Step 1 - Understand
#   Input : a list of any values.
#   Output: the same values in reverse order.
#   Ask   : reverse in place or return a new list? Can the list be empty? (yes)

# Step 2 - Examples
#   [1, 2, 3, 4] -> [4, 3, 2, 1]
#   []           -> []            (tricky: empty)
#   [7]          -> [7]           (tricky: single element)


# Step 3 - Brute force: build a new list by reading from the back.
def reverse_brute_force(nums):
    """Time O(n), extra space O(n)."""
    result = []
    for i in range(len(nums) - 1, -1, -1):
        result.append(nums[i])
    return result


# Step 4 - Optimise: the new list is wasted memory. Swap the two ends and
#          move inwards, so we reuse the input itself (O(1) extra space).

# Step 5 - Code
def reverse_in_place(nums):
    """Two pointers. Time O(n), extra space O(1)."""
    left, right = 0, len(nums) - 1
    while left < right:
        nums[left], nums[right] = nums[right], nums[left]
        left += 1
        right -= 1
    return nums


# Step 6 - Test (dry run with the examples and edge cases)
if __name__ == "__main__":
    tests = [
        [1, 2, 3, 4],
        [],
        [7],
        [1, 2, 3],        # odd length: middle element stays
        [5, 5, 5],        # all equal
        [-3, 0, 4],       # negatives and zero
    ]
    for nums in tests:
        expected = nums[::-1]
        assert reverse_brute_force(nums) == expected
        assert reverse_in_place(list(nums)) == expected
        print(nums, "->", expected)

# Step 7 - Analyse
#   reverse_brute_force : time O(n), space O(n)
#   reverse_in_place    : time O(n), space O(1)
```
